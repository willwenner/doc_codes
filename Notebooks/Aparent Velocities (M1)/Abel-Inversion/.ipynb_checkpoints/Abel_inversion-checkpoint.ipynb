```python
import numpy as np 
import scipy as sp 
import matplotlib.pyplot as plt 
```

```python
def g(y):
    return 1600*(1-y**2)

def h(y):
    return 1 - y**4

def j(y):
    return 1-3*y**2+8*y**4-6*y**6
```

```python
n = 1000
y = np.linspace(0,1,n)

mu = 0.0
sigma = .2

error = np.random.normal(mu,sigma,n)

g_noisy = j(y) + error

e_exact = #(4/(3*np.pi))*np.sqrt(1-y)*(1+2*y**2)#(3200 / np.pi) * np.sqrt(1 - y**2)
```

```python
dgdy = np.gradient(g_noisy, y)

e_noisy = np.zeros_like(y)
```

```python
for idx, y_val in enumerate(y[:-1]):
    integrand = dgdy[idx+1:]/np.sqrt(y[idx+1:]**2 - y_val**2)
    e_noisy[idx] = -(1/np.pi)*np.trapezoid(integrand, y[idx+1:])
```

```python
e_noisy[-1] = 0
```

```python
plt.plot(y,g_noisy, alpha = 0.7 , label = r'g(y) with noisy')
plt.plot(y,h(y), ls ='--', label = r'g(y)')
plt.xlabel("y")
plt.ylabel("g(y) Amplitudes")
plt.legend()
plt.grid(True, which = "major", alpha = 0.5)
plt.show()
```

```python
plt.plot(y,e_noisy, ls = '-', alpha = 0.7,label = r'$f(r)$ Numérico')
plt.plot(y,e_exact, ls = '--', label = r'$f(r)$ Exato')
plt.xlabel("r")
plt.ylabel("f(r)")
plt.legend()
plt.grid(True, which = "major", alpha = 0.5)
plt.show()
```
